```python
import pandas as pd
import numpy as np
import joblib
```

```python
df = pd.read_csv("../Datasets_For_Model_Training/Final_Merged_Dataset.csv")
df.head()
```

```python
df.info()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df.info()
```

```python
# extracting year and month
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
```

```python
df.dropna(inplace=True)
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# select the target
y = df["Electricity_Requirement"]
```

# We does require those columns

```python
X = df.drop(
    columns=[
        "Date",
        "Electricity_Requirement",
        "Total"
    ]
)
```

```python
print(X.columns)
```

```python
print("Features Shape :", X.shape)
print("Target Shape   :", y.shape)
```

```python
split = int(len(df) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]
```

```python
print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)

print("y_train :", y_train.shape)
print("y_test  :", y_test.shape)
```

# saving the datasets

```python
joblib.dump(X_train, "Demand_X_train.pkl")
joblib.dump(X_test, "Demand_X_test.pkl")

joblib.dump(y_train, "Demand_y_train.pkl")
joblib.dump(y_test, "Demand_y_test.pkl")
```
