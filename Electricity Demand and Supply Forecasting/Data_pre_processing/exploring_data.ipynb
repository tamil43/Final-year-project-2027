```python
import pandas as pd

# Read Excel
df = pd.read_csv("../Datasets_For_Model_Training/Electricity Requirement.csv")

# Convert wide format to long format
long_df = df.melt(
    id_vars=["Parameter", "STATE"],
    var_name="Date",
    value_name="Electricity_Requirement"
)

# Keep only required columns
long_df = long_df[["Date", "Electricity_Requirement"]]

# Convert Date column
long_df["Date"] = pd.to_datetime(long_df["Date"])

# Sort by date
long_df = long_df.sort_values("Date").reset_index(drop=True)

# Save
long_df.to_csv(
    "Electricity_Requirement.csv",
    index=False
)

long_df.head(30)
```

```python
long_df.dropna(inplace=True)
```
