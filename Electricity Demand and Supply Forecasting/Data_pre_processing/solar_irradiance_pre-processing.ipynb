# importing necessary libraries to do the pre-processing

```python
import numpy as np
import pandas as pd
```

# Loading the Dataset from the folder

```python
df=pd.read_csv("../Datasets/solar_irradiance.csv")
df.head()
```

# Keeping only important columns

```python
df=df.iloc[:,2:]
df.head()
```

# Extracting the columns from the datasets

```python
months=df.columns
months
```

# Doing the average for the 32 districts and making it for single year and round that value to 2 decimbal places

```python
month_avg=df.groupby("Year")[months[1:]].mean().reset_index().round(2)
month_avg
```

# converting the rows into cols for better readability and model training too

```python
final=month_avg.melt(
    id_vars="Year",
    var_name="Months",
    value_name="Solar_Irradiance"
)
```

```python
final
```

# Doing the Feature Engineering and making the Date column from the Year and Month

```python
final["Date"]=pd.to_datetime(
final["Year"].astype(str)+"-"+final["Months"],
format="%Y-%B"
)
```

```python
final=final[["Date","Solar_Irradiance"]]
final=final.sort_values("Date")
final.head()
```

```python
final.to_csv("../Datasets/solar_irradiance.csv",index=False)
```

```python
import pandas as pd

input_file = "../Datasets/solar_irradiance_from_2025.csv"
output_file = "Solar_Radiation_2025_Onwards.csv"

chunksize = 100000

first_chunk = True

for chunk in pd.read_csv(input_file, chunksize=chunksize):

    # Convert to datetime
    chunk["Data Acquisition Time"] = pd.to_datetime(
        chunk["Data Acquisition Time"],
        format="%d-%m-%Y %H:%M",
        errors="coerce"
    )

    # Filter from 01-Jan-2025 onwards
    filtered = chunk[
        chunk["Data Acquisition Time"] >= "2025-01-01"
    ]

    # Save incrementally
    filtered.to_csv(
        output_file,
        mode="w" if first_chunk else "a",
        index=False,
        header=first_chunk
    )

    first_chunk = False

print("Filtering completed successfully!")
```

```python
df=pd.read_csv("../Datasets/Solar_Radiation_2025_Onwards.csv")
df.head()
```

```python
df.info()
```

```python
df["Data Acquisition Time"] = pd.to_datetime(
    df["Data Acquisition Time"]
)
```

```python
df["Date"] = df["Data Acquisition Time"].dt.date
```

```python
daily_district = (
    df.groupby(
        ["District", "Date"]
    )["Solar Radiation (Watt/m2)"]
    .mean()
    .reset_index()
)
```

```python
daily_district.head()    
```

```python
tn_daily = (
    daily_district.groupby("Date")["Solar Radiation (Watt/m2)"]
    .mean()
    .reset_index()
)
```

```python
tn_daily.head()
```

```python
tn_daily["Date"] = pd.to_datetime(tn_daily["Date"])
```

```python
tn_daily["Year"] = tn_daily["Date"].dt.year
tn_daily["Month"] = tn_daily["Date"].dt.month
```

```python
monthly_avg = (
    tn_daily.groupby(["Year", "Month"])["Solar Radiation (Watt/m2)"]
    .mean()
    .reset_index()
)
```

```python
monthly_avg["Date"] = pd.to_datetime(
    monthly_avg[["Year", "Month"]].assign(DAY=1)
)
```

```python
monthly_avg = monthly_avg[
    ["Date", "Solar Radiation (Watt/m2)"]
]
```

```python
monthly_avg.rename(
    columns={
        "Solar Radiation (Watt/m2)": "Solar_Irradiance"
    },
    inplace=True
)
```

```python
monthly_avg.head(15)
```

```python
monthly_avg.to_csv(
    "../Datasets/TamilNadu_Solar_Irradiance_2025_Monthly.csv",
    index=False
)
```
