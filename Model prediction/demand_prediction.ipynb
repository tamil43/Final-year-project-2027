```python
import pandas as pd
import numpy as np
import joblib

from tensorflow.keras.models import load_model
```

```python
# ==========================
# Load Trained GRU Model
# ==========================

model = load_model(r"D:\Final Year Project 2027\Electricity Demand and Supply Forecasting\Model Training/Demand_GRU_Baseline_08416.keras")

# ==========================
# Load Scalers
# ==========================

X_scaler = joblib.load(r"D:\Final Year Project 2027\Electricity Demand and Supply Forecasting\Model Training/Demand_X_Scaler.pkl")
y_scaler = joblib.load(r"D:\Final Year Project 2027\Electricity Demand and Supply Forecasting\Model Training/Demand_y_Scaler.pkl")

print("Model and Scalers Loaded Successfully!")
```

```python
import pandas as pd

# ==========================================
# Future Input Data - May 2026
# ==========================================

future_data = pd.DataFrame([{
    "Humidity": 68.0,
    "Rainfall": 45.0,
    "Solar_Irradiance": 210.0,

    "Coal": 76000.0,
    "Nuclear": 3000.0,
    "Hydro": 10500.0,
    "Solar": 900.0,
    "Wind": 8500.0,
    "Small-Hydro": 180.0,
    "Bio Power": 720.0,

    "Temperature": 34.0,

    "Year": 2026,

    # May
    "Month_sin": 0.5,
    "Month_cos": -0.8660254,

    # Previous 3 Months Demand
    "Demand_Lag_1": 12922,
    "Demand_Lag_2": 12233,
    "Demand_Lag_3": 10125
}])

future_data
```

```python
future_scaled = X_scaler.transform(future_data)

print("Scaling Completed!")
```

```python
future_scaled = future_scaled.reshape(
    future_scaled.shape[0],
    1,
    future_scaled.shape[1]
)

print(future_scaled.shape)
```

```python
prediction_scaled = model.predict(future_scaled)
```

```python
prediction = y_scaler.inverse_transform(prediction_scaled)

print("=" * 50)
print("Predicted Electricity Demand")
print("=" * 50)
print(f"{prediction[0][0]:.2f} MW")
```
