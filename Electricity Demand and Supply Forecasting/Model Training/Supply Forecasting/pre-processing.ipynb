```python
import pandas as pd
import numpy as np  
```

```python
df=pd.read_excel(r"D:\\Final Year Project 2027\\Electricity Demand and Supply Forecasting\\Datasets\\Supply_Dataset.xlsx")
df.head(20)
```

```python
df = df.iloc[0:9]
df.head(10)
```

```python
# Remove State column
df = df.drop(columns=['State'])

# Set Parameter as index
df = df.set_index('Parameter')

# Transpose
df = df.T

# Reset index
df = df.reset_index()

# Rename date column
df = df.rename(columns={'index': 'Date'})

df.head()
```

```python
df.columns = [
    'Date',
    'Coal',
    'Oil_Gas',
    'Nuclear',
    'Hydro',
    'Solar',
    'Wind',
    'Small_Hydro',
    'Bio_Power',
    'Total'
]
df.head()
```

```python
# Create month number
df['Month'] = df['Date'].dt.month

# Fill missing Wind values using the median Wind generation
# for the same calendar month from other years
df['Wind'] = df['Wind'].fillna(
    df.groupby('Month')['Wind'].transform('median')
)

# Remove temporary Month column
df = df.drop(columns=['Month'])
```

```python
source_cols = [
    'Coal',
    'Oil_Gas',
    'Nuclear',
    'Hydro',
    'Solar',
    'Wind',
    'Small_Hydro',
    'Bio_Power'
]

df['Calculated_Total'] = df[source_cols].sum(axis=1)

df['Difference'] = df['Total'] - df['Calculated_Total']

df[['Date', 'Total', 'Calculated_Total', 'Difference']].head(10)
```

```python
import matplotlib.pyplot as plt

source_cols = [
    'Coal',
    'Oil_Gas',
    'Nuclear',
    'Hydro',
    'Solar',
    'Wind',
    'Small_Hydro',
    'Bio_Power'
]

for col in source_cols:
    plt.figure(figsize=(12, 4))
    plt.plot(df['Date'], df[col])
    plt.title(f'{col} Generation Over Time')
    plt.xlabel('Date')
    plt.ylabel('Generation (MU)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
source_cols = [
    'Coal',
    'Oil_Gas',
    'Nuclear',
    'Hydro',
    'Solar',
    'Wind',
    'Small_Hydro',
    'Bio_Power'
]

for col in source_cols:
    df[f'{col}_Lag_1'] = df[col].shift(1)
```

```python
df['Total_Lag_1'] = df['Total'].shift(1)
df['Total_Lag_3'] = df['Total'].shift(3)
df['Total_Lag_6'] = df['Total'].shift(6)
df['Total_Lag_12'] = df['Total'].shift(12)
```

```python
df['Total_Rolling_3'] = df['Total'].shift(1).rolling(3).mean()
df['Total_Rolling_6'] = df['Total'].shift(1).rolling(6).mean()
df['Total_Rolling_12'] = df['Total'].shift(1).rolling(12).mean()
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
import pandas as pd

# ==========================================
# 1. Final dataframe checks
# ==========================================

print("===== DATAFRAME CHECK =====")
print("Shape      :", df.shape)
print("Start Date :", df['Date'].min())
print("End Date   :", df['Date'].max())

# ==========================================
# 2. Expected 18 features
# ==========================================

expected_features = [
    'Coal_Lag_1',
    'Oil_Gas_Lag_1',
    'Nuclear_Lag_1',
    'Hydro_Lag_1',
    'Solar_Lag_1',
    'Wind_Lag_1',
    'Small_Hydro_Lag_1',
    'Bio_Power_Lag_1',
    'Year',
    'Month',
    'Quarter',
    'Total_Lag_1',
    'Total_Lag_3',
    'Total_Lag_6',
    'Total_Lag_12',
    'Total_Rolling_3',
    'Total_Rolling_6',
    'Total_Rolling_12'
]

# ==========================================
# 3. X and y
# ==========================================

X = df[expected_features]
y = df['Total']

print("\n===== X / y CHECK =====")
print("X shape :", X.shape)
print("y shape :", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

# ==========================================
# 4. Check unwanted current-month source columns
# ==========================================

unwanted = [
    'Coal',
    'Oil_Gas',
    'Nuclear',
    'Hydro',
    'Solar',
    'Wind',
    'Small_Hydro',
    'Bio_Power'
]

print("\n===== FEATURE LEAKAGE CHECK =====")
leakage_features = [col for col in unwanted if col in X.columns]

if len(leakage_features) == 0:
    print("PASS: No current-month source-generation columns in X")
else:
    print("FAIL: Found unwanted columns:", leakage_features)

# ==========================================
# 5. Missing-value check
# ==========================================

print("\n===== MISSING VALUE CHECK =====")
print(X.isnull().sum())

# ==========================================
# 6. Train / test split
# ==========================================

X_train = X.iloc[:-24]
X_test = X.iloc[-24:]

y_train = y.iloc[:-24]
y_test = y.iloc[-24:]

print("\n===== TRAIN / TEST CHECK =====")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

print(
    "Training period:",
    df['Date'].iloc[0],
    "to",
    df['Date'].iloc[len(X_train)-1]
)

print(
    "Testing period :",
    df['Date'].iloc[len(X_train)],
    "to",
    df['Date'].iloc[-1]
)
```
