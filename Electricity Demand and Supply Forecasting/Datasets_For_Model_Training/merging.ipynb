```python
import pandas as pd
from functools import reduce

# Read all CSV files
electricity = pd.read_csv("Electricity Requirement.csv")
humidity = pd.read_csv("Humidity Datasets.csv")
rainfall = pd.read_csv("Rainfall datasets.csv")
supply=pd.read_csv("supply.csv")
solar = pd.read_csv("solar_irradiance.csv")
temperature = pd.read_csv("Temperature Dataset.csv")

# Convert Date column to datetime
datasets = [
    electricity,
    humidity,
    rainfall,
    supply,
    solar,
    temperature
]

for df in datasets:
    df["Date"] = pd.to_datetime(df["Date"])

# Merge all datasets on Date
merged_df = reduce(
    lambda left, right: pd.merge(
        left,
        right,
        on="Date",
        how="inner"
    ),
    datasets
)

# Sort by Date
merged_df = merged_df.sort_values("Date").reset_index(drop=True)

# Save merged dataset
merged_df.to_csv("Final_Merged_Dataset.csv", index=False)
merged_df.head(15)
```

```python
# Check the Date column format in all datasets

datasets = {
    "Electricity": electricity,
    "Humidity": humidity,
    "Rainfall": rainfall,
    "Supply": supply,
    "Solar": solar,
    "Temperature": temperature
}

for name, df in datasets.items():
    print("=" * 60)
    print(f"{name} Dataset")

    # Check if Date column exists
    if "Date" in df.columns:
        print("\nDate Data Type:")
        print(df["Date"].dtype)

        print("\nFirst 10 Dates:")
        print(df["Date"].head(10).tolist())

        print("\nLast 10 Dates:")
        print(df["Date"].tail(10).tolist())

        print("\nTotal Rows:", len(df))

    else:
        print("No 'Date' column found!")
```

```python
# Fix the wrongly encoded monthly dates

humidity["Date"] = pd.to_datetime({
    "year": humidity["Date"].dt.year,
    "month": humidity["Date"].dt.day,
    "day": 1
})

rainfall["Date"] = pd.to_datetime({
    "year": rainfall["Date"].dt.year,
    "month": rainfall["Date"].dt.day,
    "day": 1
})

solar["Date"] = pd.to_datetime({
    "year": solar["Date"].dt.year,
    "month": solar["Date"].dt.day,
    "day": 1
})

temperature["Date"] = pd.to_datetime({
    "year": temperature["Date"].dt.year,
    "month": temperature["Date"].dt.day,
    "day": 1
})
```
