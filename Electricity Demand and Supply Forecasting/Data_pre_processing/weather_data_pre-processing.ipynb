```python
import numpy as np
import pandas as pd
```

# We are Pre-processing the Humidity Datasets

```python
df = pd.read_csv(
    "../Datasets/Humidity Datasets.csv",
    skiprows=21
)

df.head()
```

```python
months=df.columns
```

```python
monthly_avg = df.groupby("YEAR")[months[3:]].mean().reset_index()
```

```python
monthly_avg.drop("ANN",axis=1,inplace=True)
```

```python
humidity = monthly_avg.melt(
    id_vars="YEAR",
    var_name="Month",
    value_name="Humidity"
)
```

```python
humidity
```

```python
humidity["Date"] = pd.to_datetime(
    humidity["YEAR"].astype(str) + "-" + humidity["Month"],
    format="%Y-%b"
)
```

```python
humidity
```

```python
humidity = humidity[["Date", "Humidity"]]
humidity
```

```python
humidity = humidity.sort_values("Date").reset_index(drop=True)
```

```python
humidity
```

```python
humidity["Humidity"] = humidity["Humidity"].round(2)
humidity
```

```python
humidity.to_csv("../Datasets/Humidity Datasets.csv",index=False)
```

# 👆👆 we are converted to the structured datasets of humidity

# 👇 Going to process the Rainfall Datasets

```python
df=pd.read_csv("../Datasets/Rainfall datasets.csv")
df.head()
```

```python
df=df.iloc[:,1:]
df.head()
```

```python
monthly_avg = df.groupby("YEAR")[
    ["JAN","FEB","MAR","APR","MAY","JUN",
     "JUL","AUG","SEP","OCT","NOV","DEC"]
].mean().reset_index()
```

```python
monthly_avg
```

```python
rainfall = monthly_avg.melt(
    id_vars="YEAR",
    var_name="Month",
    value_name="Rainfall"
)
rainfall
```

```python
rainfall["Date"] = pd.to_datetime(
    rainfall["YEAR"].astype(str) + "-" + rainfall["Month"],
    format="%Y-%b"
)
rainfall
```

```python
rainfall = rainfall[["Date", "Rainfall"]]
rainfall
```

```python
rainfall = rainfall.sort_values("Date").reset_index(drop=True)
rainfall
```

# 👇Replaced existing rainfall file with structured one

```python
rainfall.to_csv("../Datasets/Rainfall datasets.csv",index=False)
```

# 👇 Pre-processing the Temperature datasets

```python
df=pd.read_csv("../Datasets/Temperature Dataset.csv",skiprows=21)
df.head()
```

```python
df=df.iloc[:,1:]
df.head()
```

```python
months=df.columns
```

```python
monthly_avg=df.groupby("YEAR")[months[3:]].mean().reset_index()
monthly_avg.drop("ANN",axis=1,inplace=True)
```

```python
final=monthly_avg.melt(
    id_vars="YEAR",
    var_name="Month",
    value_name="Temperature"
)
```

```python
final
```

```python
final["Date"] = pd.to_datetime(
    final["YEAR"].astype(str) + "-" + final["Month"],
    format="%Y-%b"
)
final
```

```python
final=final[["Date","Temperature"]]
final["Temperature"]=final["Temperature"].round(2)
final
```

```python
final.to_csv("../Datasets/Temperature Dataset.csv",index=False)
```
