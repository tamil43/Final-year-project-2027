```python
supply=pd.read_excel("../Datasets/Supply dataset.xlsx")
supply.head()
```

```python
supply = supply.set_index("Parameter")

# Transpose (rows ↔ columns)
supply = supply.T

# Convert index into a column
supply = supply.reset_index()

# Rename the first column
supply = supply.rename(columns={"index": "Date"})

# Remove the name of the columns index (optional but recommended)
supply.columns.name = None

# Convert Date column to datetime
supply["Date"] = pd.to_datetime(supply["Date"], format="%Y-%m")

# Display
supply.head()
```

```python
cols=supply.columns
supply=supply.drop(columns=cols[-1])
supply.head()
```

```python
supply.to_csv("../Datasets_For_Model_Training/Supply dataset.csv",index=False)
```

```python
import pandas as pd
import os
from functools import reduce

# Folder containing all datasets
folder_path = "../Datasets_For_Model_Training/"

# Read all CSV files
dfs = []

for file in os.listdir(folder_path):
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(folder_path, file), parse_dates=["Date"])
        dfs.append(df)

# Merge all datasets on the Date column
merged_df = reduce(
    lambda left, right: pd.merge(left, right, on="Date", how="inner"),
    dfs
)

# Display
print(merged_df.head())
print("Shape:", merged_df.shape)

# Save the merged dataset
merged_df.to_csv("../Datasets/final_merged_dataset.csv", index=False)

print("Merged successfully!")
```
