```python
import pandas as pd
import glob
import os
```

```python
folder_path = r"../Datasets/solar irradiance datasets"
```

```python
excel_files = glob.glob(os.path.join(folder_path, "*.xlsx"))

print("Total Files:", len(excel_files))
```

```python
df_list = []

for file in excel_files:
    df = pd.read_excel(file)
    df_list.append(df)

merged_df = pd.concat(df_list, ignore_index=True)

print("Merged Successfully!")
print("Rows :", merged_df.shape[0])
print("Columns :", merged_df.shape[1])
```

```python
merged_df.head()
```

```python
month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

merged_df["Month"] = pd.Categorical(
    merged_df["Month"],
    categories=month_order,
    ordered=True
)
```

```python
pivot_df = merged_df.pivot_table(
    index=["State", "District", "Year"],
    columns="Month",
    values="Insolation (in kWh/m²)"
).reset_index()
```

```python
pivot_df["Year"] = pivot_df["Year"].astype(int)
```
