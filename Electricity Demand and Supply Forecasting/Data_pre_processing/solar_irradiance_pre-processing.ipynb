# importing necessary libraries to do the pre-processing

```python
import numpy as np
import pandas as pd
```

# Loading the Dataset from the folder

```python
df=pd.read_csv("../Datasets/solar_irradiance.csv")
df.head()
```

# Keeping only important columns

```python
df=df.iloc[:,2:]
df.head()
```

# Extracting the columns from the datasets

```python
months=df.columns
months
```

# Doing the average for the 32 districts and making it for single year and round that value to 2 decimbal places

```python
month_avg=df.groupby("Year")[months[1:]].mean().reset_index().round(2)
month_avg
```

# converting the rows into cols for better readability and model training too

```python
final=month_avg.melt(
    id_vars="Year",
    var_name="Months",
    value_name="Solar_Irradiance"
)
```

```python
final
```

# Doing the Feature Engineering and making the Date column from the Year and Month

```python
final["Date"]=pd.to_datetime(
final["Year"].astype(str)+"-"+final["Months"],
format="%Y-%B"
)
```

```python
final=final[["Date","Solar_Irradiance"]]
final=final.sort_values("Date")
final.head()
```

```python
final.to_csv("../Datasets/solar_irradiance.csv",index=False)
```
