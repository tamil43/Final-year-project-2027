```python
import pandas as pd
from functools import reduce

# Read all CSV files
electricity = pd.read_csv("Electricity Requirement.csv")
humidity = pd.read_csv("Humidity Datasets.csv")
rainfall = pd.read_csv("Rainfall datasets.csv")
solar = pd.read_csv("solar_irradiance.csv")
supply = pd.read_csv("Supply dataset.csv")
temperature = pd.read_csv("Temperature Dataset.csv")

# Convert Date column to datetime
datasets = [
    electricity,
    humidity,
    rainfall,
    solar,
    supply,
    temperature
]

for df in datasets:
    df["Date"] = pd.to_datetime(df["Date"])

# Merge all datasets on Date
merged_df = reduce(
    lambda left, right: pd.merge(
        left,
        right,
        on="Date",
        how="inner"
    ),
    datasets
)

# Sort by Date
merged_df = merged_df.sort_values("Date").reset_index(drop=True)

# Save merged dataset
merged_df.to_csv("Final_Merged_Dataset.csv", index=False)
merged_df.head(15)
```
